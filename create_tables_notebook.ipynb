```python
import pandas as pd
```

```python
df = pd.read_csv('Data.csv')
```

```python
df = df.drop(columns=["Row ID"])
df = df.drop(columns=["Country"])
```

## 1. Convert date columns to datetime

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True
)
```

## 2. Drop the 11 Null

```python
df.dropna(subset=['Postal Code'], inplace=True)
```

## 3. Feature Engineering: Calculate shipping duration in days

```python
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
df.insert(
    df.columns.get_loc("Ship Date") + 1,
    "Shipping Days",
    df.pop("Shipping Days")
)
```

## First page - Supply chain Dashboard

```python
# 1. Calculate Key Performance Indicators (KPIs)
average_shipping_days = df['Shipping Days'].mean()
maximum_shipping_days = df['Shipping Days'].max()

print("--- Supply Chain Key Metrics ---")
print(f"Average Shipping Time: {average_shipping_days:.2f} days")
print(f"Maximum Shipping Delay (Worst Case): {maximum_shipping_days} days")
print("\n" + "="*40 + "\n")

# 2. Analyze Shipping Days by Ship Mode
ship_mode_analysis = df.groupby('Ship Mode')['Shipping Days'].mean().sort_values(ascending=False)

print("--- Average Shipping Days by Ship Mode ---")
print(ship_mode_analysis)
```

## 👥 Marketing & Customer Analytics Insights

### 1. Customer Lifetime Value (CLV) & "Whales" Identification
* **The Ultimate Whale:** **Sean Miller** (`SM-20320`) is the highest-value customer for the company, with a massive lifetime spend of **$25,043.05**. 
* **Top Tier Contributors:** Customers like **Tamara Chand ($19,052.22)** and **Raymond Buch ($15,117.34)** follow closely. These top 5 individuals represent critical high-value accounts that require dedicated relationship management and exclusive VIP offers to ensure retention.

```python
# 1. Calculate Customer Lifetime Value (CLV) as Total Sales per Customer
clv_df = df.groupby(['Customer ID', 'Customer Name'])['Sales'].sum().reset_index()
clv_df.rename(columns={'Sales': 'CLV_Total_Sales'}, inplace=True)
```

### 2. Customer Segmentation Strategy (`pd.cut` / Logic)
* **The Revenue Backbone (Regulars):** The majority of the customer base falls into the **Regular** segment (**501 customers** spending between $1k and $5k). This is a healthy sign, showing a strong, stable middle-class customer layer that generates consistent revenue.
* **The Core Asset (VIPs):** There are **113 VIP customers** (spending > $5k). Although they represent a smaller portion of the total customer count, their high CLV likely contributes disproportionately to total profitability. 
* **The Growth Opportunity (Occasionals):** There are **179 Occasional customers** (spending < $1k). Marketing should target this specific segment with re-engagement campaigns, bundle discounts, or first-purchase incentives to convert them into "Regular" buyers.

```python
# 2. Customer Segmentation based on CLV thresholds
# VIP: > 5000 | Regular: 1000 - 5000 | Occasional: < 1000
def assign_segment(sales):
    if sales > 5000:
        return 'VIP'
    elif 1000 <= sales <= 5000:
        return 'Regular'
    else:
        return 'Occasional'

clv_df['Customer_Segment'] = clv_df['CLV_Total_Sales'].apply(assign_segment)

# 3. Count how many customers are in each segment
segment_counts = clv_df['Customer_Segment'].value_counts()

print("--- Top 5 Whales (Highest CLV) ---")
print(clv_df.sort_values(by='CLV_Total_Sales', ascending=False).head(5))
print("\n" + "="*50 + "\n")

print("--- Customer Segment Counts ---")
print(segment_counts)
```

### 🎯 Product Preference Analysis by Segment

#### 1. VIP Segment Preferences (The Tech Buyers)
* **High-Ticket Technology:** The **VIP** segment heavily drives **Technology** sales ($394,226.71). 
* **Top Sub-Categories:** Their absolute favorites are **Machines ($138,012.75)** and **Phones ($114,545.55)**, followed by **Chairs ($111,135.64)**. This proves that VIPs are purchasing high-value infrastructure and heavy enterprise equipment. Marketing should target them with tech upgrades and premium enterprise contracts.

#### 2. Regular Segment Preferences (The Infrastructure Backbone)
* **Balanced Core Buying:** The **Regular** segment spends almost equally across all categories, with **Furniture ($459,023.07)** leading slightly.
* **Top Sub-Categories:** Their top purchases are **Chairs ($199,089.30)** and **Phones ($197,590.26)**. This segment represents steady office setups and standard business operations.

#### 3. Occasional Segment Preferences (The Essentials Buyers)
* **Low-Cost Office Supplies:** **Occasional** buyers focus primarily on **Office Supplies ($42,606.81)**, which are low-ticket, day-to-day items.
* **Top Sub-Categories:** While **Phones** and **Chairs** appear in their top 3, their overall volumes are very low. They mostly come to buy basic essentials. Cross-selling Tech bundles could help upgrade them to the Regular tier.

```python
# 1. Merge customer segments back into the main DataFrame
df_segmented = df.merge(clv_df[['Customer ID', 'Customer_Segment']], on='Customer ID', how='left')

# 2. Analyze Product Categories by Segment
category_pref = df_segmented.groupby(['Customer_Segment', 'Category'])['Sales'].sum().reset_index()
category_pref = category_pref.sort_values(['Customer_Segment', 'Sales'], ascending=[True, False])

print("--- Category Sales by Customer Segment ---")
print(category_pref)
print("\n" + "="*50 + "\n")

# 3. Analyze Top 3 Sub-Categories for each Segment to find deep preferences
print("--- Top 3 Sub-Categories by Segment ---")
subcat_pref = df_segmented.groupby(['Customer_Segment', 'Sub-Category'])['Sales'].sum().reset_index()

for segment in ['VIP', 'Regular', 'Occasional']:
    print(f"\n[ {segment} Customer Preferences ]")
    top_3_sub = subcat_pref[subcat_pref['Customer_Segment'] == segment].sort_values(by='Sales', ascending=False).head(3)
    print(top_3_sub[['Sub-Category', 'Sales']])
```

```python
import urllib
import pandas as pd
from sqlalchemy import create_engine, text

# ==========================================
# 1. YOUR DATAFRAME
# ==========================================
# Assuming your dataframe variable is named 'df'
# df = pd.read_csv(...) or wherever your 9789 rows are stored

# ==========================================
# 2. CONFIGURING CONNECTION TO YOUR SQL SERVER
# ==========================================
# Configured specifically to match your IDE header: MOKA\SQLEXPRESS
connection_string = (
    r"DRIVER={ODBC Driver 17 for SQL Server};"
    r"SERVER=MOKA\SQLEXPRESS;"
    r"DATABASE=RetailSupplyChain;"
    r"Trusted_Connection=yes;"
)

params = urllib.parse.quote_plus(connection_string)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

print("Connection established successfully!")

# ==========================================
# 3. PUSH RAW DATA TO A STAGING TABLE
# ==========================================
print("Uploading raw DataFrame to staging area...")
# This automatically uploads your pandas data into SQL Server
df.to_sql("Staging_Superstore", con=engine, if_exists="replace", index=False)
print("Staging data uploaded.")

# ==========================================
# 4. RUN PIPELINE TO POPULATE RELATION TABLES
# ==========================================
# We process the data sequentially to respect foreign key constraints
migration_sql = """
-- Populate Customers
INSERT INTO Customers (CustomerID, CustomerName, Segment)
SELECT [Customer ID], [Customer Name], [Segment]
FROM (
    SELECT [Customer ID], [Customer Name], [Segment],
           ROW_NUMBER() OVER (PARTITION BY [Customer ID] ORDER BY [Order Date] DESC) as rn
    FROM Staging_Superstore
) t WHERE t.rn = 1 
  AND [Customer ID] NOT IN (SELECT CustomerID FROM Customers);

-- Populate Locations (and strip out pandas float decimals like '.0' from postal codes)
INSERT INTO Locations (PostalCode, City, State, Region)
SELECT DISTINCT 
    CASE 
        WHEN CHARINDEX('.', CAST([Postal Code] AS VARCHAR)) > 0 
        THEN LEFT(CAST([Postal Code] AS VARCHAR), CHARINDEX('.', CAST([Postal Code] AS VARCHAR)) - 1)
        ELSE ISNULL(CAST([Postal Code] AS VARCHAR), '00000') 
    END,
    [City], [State], [Region]
FROM Staging_Superstore s
WHERE NOT EXISTS (
    SELECT 1 FROM Locations l 
    WHERE l.City = s.City AND l.State = s.State AND l.Region = s.Region
);

-- Populate Products
INSERT INTO Products (ProductID, ProductName, Category, SubCategory)
SELECT [Product ID], [Product Name], [Category], [Sub-Category]
FROM (
    SELECT [Product ID], [Product Name], [Category], [Sub-Category],
           ROW_NUMBER() OVER (PARTITION BY [Product ID] ORDER BY [Order Date] DESC) as rn
    FROM Staging_Superstore
) t WHERE t.rn = 1
  AND [Product ID] NOT IN (SELECT ProductID FROM Products);

-- Populate Orders Header Table
INSERT INTO Orders (OrderID, CustomerID, LocationID, OrderDate, ShipDate, ShipMode, ShippingDays)
SELECT [Order ID], [Customer ID], LocationID, [Order Date], [Ship Date], [Ship Mode], [Shipping Days]
FROM (
    SELECT 
        s.[Order ID], s.[Customer ID], l.LocationID, 
        CAST(s.[Order Date] AS DATE) AS [Order Date], 
        CAST(s.[Ship Date] AS DATE) AS [Ship Date], 
        s.[Ship Mode], s.[Shipping Days],
        ROW_NUMBER() OVER (PARTITION BY s.[Order ID] ORDER BY s.[Order Date] DESC) as rn
    FROM Staging_Superstore s
    JOIN Locations l ON l.City = s.City 
                    AND l.State = s.State 
                    AND l.Region = s.Region
) t WHERE t.rn = 1
  AND [Order ID] NOT IN (SELECT OrderID FROM Orders);

-- Populate OrderDetails (Line items)
INSERT INTO OrderDetails (OrderID, ProductID, Sales)
SELECT [Order ID], [Product ID], [Sales]
FROM Staging_Superstore;

-- Clean up the staging table when done
DROP TABLE Staging_Superstore;
"""

print("Processing relational migrations...")
with engine.begin() as connection:
    connection.execute(text(migration_sql))

print("ETL complete! Go back to SQL Server and re-run your SELECT statement.")
```
